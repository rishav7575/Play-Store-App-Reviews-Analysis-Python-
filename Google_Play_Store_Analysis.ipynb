#Fixing the Size

```python
playstore_df['Size'].unique()
```

```python
def fix_size(s):
  try:
    if s.endswith('M'):
      return float(s[:-1])
    elif s.endswith('k'):
      return (float(s[:-1]) / 1024)
    elif s == '1,000+':
      return 1.0
    else:
      return np.nan
  except:
      return np.nan
```

Demo

```python
fix_size('98M')
```

```python
fix_size('45k')
```

```python
fix_size('demo')
```

#Fixing all elements in Size columns

```python
playstore_df['Size']= playstore_df['Size'].apply(fix_size)
```

```python
playstore_df['Size'].unique()
```

```python
playstore_df.info()
```

```python
playstore_df['Price'].unique()
```

```python
def fix_Price(A):
  try:
    if A.startswith('$'):
      return float(A[1:])
    else:
      return 0.0
  except:
    return np.nan
```

```python
playstore_df.info()
```

```python
fix_Price('Everyone')
```

```python
playstore_df['Price']=playstore_df['Price'].apply(fix_Price)
```

```python
def lower_category(B):
  return B.title()
```

```python
lower_category('AUTO_AND_VEHICLES')
```

```python

playstore_df['Category'] =playstore_df['Category'].apply(lower_category)
```

```python
playstore_df['Installs'].unique()
```

```python
playstore_df['Installs'] = playstore_df.Installs.str.replace(",","")
playstore_df['Installs'] = playstore_df.Installs.str.replace("+","")
playstore_df['Installs'] = playstore_df['Installs'].astype(float)
playstore_df['Installs'].dtype
```

```python
playstore_df['Installs'].unique()
```

```python
playstore_df.info()
```

```python
playstore_df['Android Ver'].unique()
```

```python
def correction(y):
  try:
    if y == 'Varies with device':
      return np.nan
    elif y == '4.4W and up':
      return y.replace('W','')
    else:
      return y
  except:
      return y
```

```python
playstore_df['Android Ver'] = playstore_df['Android Ver'].apply(correction)
```

```python
playstore_df['Android Ver'].unique()
```

#Relation between app category and app price

```python
fig, ax = plt.subplots()
fig.set_size_inches(10, 5)

popular_app_cats = playstore_df[playstore_df.Category.isin(['Game', 'Family', 'Photography',
                                            'Medical', 'Tools', 'Finance',
                                            'Lifestyle','Business'])]

ax = sns.stripplot(x = popular_app_cats["Price"], y = popular_app_cats["Category"], jitter=True, linewidth=1,  palette = 'Set2' )
ax.set_title('App pricing trend across categories')
```

#Sentiment analysis of user reviews

```python
merged_df.head(3)
```

#Filtering out "junk" apps

```python
apps_under_100 = popular_app_cats[popular_app_cats["Price"]<100]

fig, ax = plt.subplots()
fig.set_size_inches(15, 8)

ax = sns.stripplot(x = "Price", y ="Category", data = apps_under_100, jitter = True, linewidth = 1, palette = 'Set1')
ax.set_title('App pricing trend across categories after filtering for junk apps')
```
